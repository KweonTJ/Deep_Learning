## 2.1 퍼셉트론이란?
- 다수의 신호를 입력으로 받아 하나의 신호를 출력함.
- 두 신호 입력신호에 각각 가중치를 곱하고 편향을 더한 출력값을 활성화 함수에 넣어 출력함.
- 퍼셉트론은 복수의 입력 신호 각각에 고유 가중치를 부여. 가중치는 각 신호가 결과에 주는 영향력을 조절하는 요소로 작용. 클수록 신호 중요도 증가.

## 2.3 퍼셉트론 구현하기

```python
def AND(x1, x2):
    w1, w2, theta = 0.5, 0.5, 0.7
    tmp = x1*w1 + x2*w2
    if tmp <= theta:
        return 0
    elif tmp > theta:
        return 1

print(AND(0, 0))
print(AND(1, 0))
print(AND(0, 1))
print(AND(1, 1))
```

```python
def AND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

print(AND(0, 0))
print(AND(1, 0))
print(AND(0, 1))
print(AND(1, 1))
```

```python
def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

print(NAND(0, 0))
print(NAND(1, 0))
print(NAND(0, 1))
print(NAND(1, 1))
```

```python
def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.2
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

print(OR(0, 0))
print(OR(1, 0))
print(OR(0, 1))
print(OR(1, 1))
```

## 2.4 퍼셉트론의 한계
- 퍼셉트론 이용시 3가지의 논리 회루 구현.
- XOR 게이트는 배타적 논리합 회로. 퍼셉트론으로 구현 불가.

## 2.5 다층 퍼셉트론 출동
- 퍼셉트론의 아름다움은 층을 쌓아 다층 퍼셉트론 구현
- XOR 게이트는 NAND, OR 게이트에 각각의 원소가 입력하고, 해당 게이트 출력값을 AND 게이트 입력으로 변환하여 최종 출력

    - 다층 퍼셉트론 : 층이 여러 개인 퍼셉트론
    - 단층 퍼셉트론으로는 표현하지 못한 것을 층을 추가해 구현

```python
def XOR(x1, x2):
    s1 = NAND(x1, x2)
    s2 = OR(x1, x2)
    y = AND(s1, s2)
    return y

print(XOR(0, 0))
print(XOR(1, 0))
print(XOR(0, 1))
print(XOR(1, 1))
```

## 2.6 NAND에서 컴퓨터까지
- 다층퍼셉트론은 복잡한 회로 제작이 가능함.
- 가산기, 인코더, 패리티 검사 회로도 퍼셉트론으로 표현 가능.
- 이론상 2층 퍼셉트론이면 컴퓨터를 제작할 수 있음.
    - 비선형 시그모이드 함수를 활성화 함수로 이용하면 임의의 함수를 표현할 수 있다는 사실 증명
    - 실제로 NAND등의 저수준 소자에서 시작해 컴퓨터에 필요한 부품을 단계적으로 제작.
    - 제작 순서 : AND, OR -> 반가산기, 전가산기 -> ALU -> CPU
    - 퍼셉트론으로 표현하는 컴퓨터도 여러 층을 다시 층층이 겹치는 구조로 만드는 방향이 자연스러움
