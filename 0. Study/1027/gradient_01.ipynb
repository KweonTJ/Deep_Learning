```python
import numpy as np
import matplotlib.pyplot as plt

def numerical_diff(f, x):
    h = 1e-4  # A small number
    return (f(x + h) - f(x - h)) / (2 * h)

def function_1(x):
    return 0.01 * x ** 2 + 0.1 * x

x = np.arange(0.0, 20.0, 0.1)
y = function_1(x)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.plot(x, y)
plt.show()
```

```python
def function_2(x):
    return x[0]**2 + x[1]**2

import numpy as np
import matplotlib.pyplot as plt

# 그래프 범위
x0 = np.linspace(-3, 3, 100)
x1 = np.linspace(-3, 3, 100)

X0, X1 = np.meshgrid(x0, x1)
Y = function_2(np.array([X0, X1]))

# 기울기 계산
gx, gy = np.meshgrid(
    np.linspace(-3, 3, 13),
    np.linspace(-3, 3, 13)
)

grad = np.array([
    numerical_gradient(function_2, np.array([a, b], dtype=float))
    for a, b in zip(gx.ravel(), gy.ravel())
])

U = grad[:, 0].reshape(gx.shape)
V = grad[:, 1].reshape(gy.shape)

fig = plt.figure(figsize=(12, 5))

# 3D 함수 그래프
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(X0, X1, Y, cmap='viridis')

ax1.set_xlabel('x0')
ax1.set_ylabel('x1')
ax1.set_zlabel('y')
ax1.set_zlim(0, 18)
ax1.set_title('Function Surface')

# 2D 기울기 벡터 그래프
ax2 = fig.add_subplot(122)

ax2.contour(X0, X1, Y, levels=10, cmap='viridis')
ax2.quiver(gx, gy, U, V, color='blue')

ax2.set_xlim(-3, 3)
ax2.set_ylim(-3, 3)
ax2.set_aspect('equal')
ax2.set_xlabel('x0')
ax2.set_ylabel('x1')
ax2.set_title('Numerical Gradient')

plt.tight_layout()
plt.show()
```
