```python
import numpy as np 

a = np.array([0.3, 2.9, 4.0])

exp_a = np.exp(a)
print(exp_a)
```

```python
sum_exp_a = np.sum(exp_a)
print(sum_exp_a)
```

```python
y = exp_a / sum_exp_a
print(y)
```

```python
def softmax(a):
    exp_a = np.exp(a)
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a
    return y
```

```python
def softmax(a):
    c = np.max(a)
    exp_a = np.exp(a - c)  # Prevent overflow by subtracting the max
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a

    return y
```

```python
a = np.array([0.3, 2.9, 4.0])
y = softmax(a)
print(y)
np.sum(y)  # This should be 1.0
```

```python
import math

def softmax(a):
    exp_a = [math.exp(i) for i in a]
    sum_exp_a = sum(exp_a)

    return [i / sum_exp_a for i in exp_a]

x = [0.3, 2.9, 4.0]
result = softmax(x)
print(result)
```
